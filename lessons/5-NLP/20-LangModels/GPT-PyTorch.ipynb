```python
from transformers import pipeline, set_seed
import torch

# Use GPT-2 (better than original openai-gpt)
model_name = 'gpt2'  # Options: 'gpt2', 'gpt2-medium', 'gpt2-large', 'gpt2-xl'

print(f"Loading model: {model_name}")
generator = pipeline('text-generation', model=model_name)
set_seed(42)  # For reproducibility
print("Model loaded successfully!")
```

## 2. Basic Text Generation

GPT generates text by predicting the next word given the previous context. Let's try some basic generation:

```python
# Basic text generation
prompt = "Hello! I am a neural network, and I want to say that"
results = generator(prompt, max_length=100, num_return_sequences=3)

for i, result in enumerate(results, 1):
    print(f"\n--- Generated Text {i} ---")
    print(result['generated_text'])
```

## 4. Text Sampling Strategies

There are different strategies for selecting the next word during generation:

### 4.1 Greedy Search (Default)
Always selects the most probable next word. Can lead to repetitive text.

```python
prompt = "It was early evening when I came back from work. I usually work late, but this time it was an exception. When I entered the room, I saw"

# Greedy search (default)
result = generator(prompt, max_length=100, num_return_sequences=1)
print("Greedy Search:")
print(result[0]['generated_text'])
```

### 4.2 Beam Search
Explores multiple generation paths and selects the one with highest overall probability. Can produce more coherent text.

```python
# Beam search with no repeat n-gram penalty
results = generator(
    prompt, 
    max_length=100, 
    num_return_sequences=3, 
    num_beams=5,
    no_repeat_ngram_size=2,
    early_stopping=True
)

print("Beam Search Results:")
for i, result in enumerate(results, 1):
    print(f"\n--- Result {i} ---")
    print(result['generated_text'])
```

### 4.3 Sampling with Temperature
Selects next word probabilistically. Temperature controls randomness:
- Low temperature (0.1-0.5): More deterministic, focused
- High temperature (0.8-1.5): More random, creative

```python
# Low temperature - more focused
print("Temperature = 0.5 (more focused):")
results = generator(prompt, max_length=80, do_sample=True, temperature=0.5, num_return_sequences=2)
for result in results:
    print(result['generated_text'])
    print("-" * 50)
```

```python
# High temperature - more creative
print("Temperature = 1.2 (more creative):")
results = generator(prompt, max_length=80, do_sample=True, temperature=1.2, num_return_sequences=2)
for result in results:
    print(result['generated_text'])
    print("-" * 50)
```

### 4.4 Top-K and Top-P Sampling

- **Top-K**: Only consider the K most probable next words
- **Top-P (nucleus sampling)**: Consider the smallest set of words whose cumulative probability exceeds P

```python
# Top-K sampling
print("Top-K Sampling (k=50):")
results = generator(
    prompt, 
    max_length=80, 
    do_sample=True, 
    top_k=50,
    num_return_sequences=2
)
for result in results:
    print(result['generated_text'])
    print("-" * 50)
```

```python
# Top-P (nucleus) sampling
print("Top-P Sampling (p=0.9):")
results = generator(
    prompt, 
    max_length=80, 
    do_sample=True, 
    top_p=0.9,
    num_return_sequences=2
)
for result in results:
    print(result['generated_text'])
    print("-" * 50)
```

```python
# Combined: Top-K + Top-P + Temperature (recommended)
print("Combined Sampling (top_k=50, top_p=0.95, temp=0.8):")
results = generator(
    prompt, 
    max_length=80, 
    do_sample=True, 
    top_k=50,
    top_p=0.95,
    temperature=0.8,
    num_return_sequences=2
)
for result in results:
    print(result['generated_text'])
    print("-" * 50)
```

## 5. Understanding Perplexity

**Perplexity** measures how well a language model predicts text. Lower perplexity = better model.

$$\text{Perplexity}(W) = \sqrt[N]{\frac{1}{P(W_1,...,W_N)}}$$

A model with low perplexity is less "surprised" by real text.

```python
import math
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# Load model and tokenizer for perplexity calculation
model = GPT2LMHeadModel.from_pretrained(model_name)
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model.eval()

def calculate_perplexity(text):
    """Calculate perplexity of a text."""
    encodings = tokenizer(text, return_tensors='pt')
    
    with torch.no_grad():
        outputs = model(**encodings, labels=encodings['input_ids'])
        loss = outputs.loss
        
    perplexity = math.exp(loss.item())
    return perplexity

# Test with different sentences
sentences = [
    "The quick brown fox jumps over the lazy dog.",
    "Artificial intelligence is transforming the world.",
    "Can it does what?"  # Grammatically incorrect
]

print("Perplexity Analysis:")
print("=" * 60)
for sent in sentences:
    ppl = calculate_perplexity(sent)
    print(f"Text: {sent}")
    print(f"Perplexity: {ppl:.2f}")
    print("-" * 60)
```
